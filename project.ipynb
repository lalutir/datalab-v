### Samenvatting van de Opdracht

**Context & Opdrachtgever**

* **Opdrachtgever:** Lidi Smart Solutions, met contactpersoon Lidia Dimitrova.


* **Achtergrond:** De opdracht is onderdeel van het *AI4AgriBizAfrica* concept, dat zich richt op het verbeteren van klimaat- en weermodellen in Afrika om boeren te helpen bij het omgaan met droogtes en overstromingen.



**Het Probleem**

* Vroege signalen van klimaatrampen zijn vaak verborgen in complexe atmosferische patronen en grote hoeveelheden data worden onvoldoende benut.


* De kernvraag is hoe je met open klimaatdata een machine learning model kunt bouwen dat de kans op droogte en overstromingen in specifieke Afrikaanse regio's voorspelt.



**De Dataset**

* Je maakt gebruik van open-access ERA5 reanalysis data van Copernicus, opgeslagen in een Microsoft Azure cloudomgeving.


* De data beslaat het hele Afrikaanse continent, is op uurbasis gemeten, en is verdeeld over twee periodes: 1998-2003 en 2018-2023.


* De dataset bevat 6 atmosferische parameters: temperatuur, dauwpunt, twee windcomponenten, luchtdruk op zeeniveau en totale neerslag. Er is geen sprake van privacygevoelige informatie.



**Doelen & Deliverables**

* Een werkend *proof-of-concept* model voor het voorspellen van droogte (classificatie) en overstromingsrisico (classificatie of regressie).


* Literatuuronderzoek naar welke variabelen en *features* de grootste invloed hebben op deze weersomstandigheden.


* Vergelijking en evaluatie van meerdere ML-technieken (zoals Logistic Regression, Random Forest, XGBoost) met behulp van Python.


* Duidelijke visualisaties en rapportage voor non-technische stakeholders.


* Een reflectie op de ethische en maatschappelijke aspecten (zoals *bias* en regionale eerlijkheid) van klimaatrisicomodellering.

### Plan van Aanpak

Om dit project succesvol te doorlopen, kun je het project het beste opdelen in de volgende fases:

**Fase 1: Domeinkennis & Probleemdefinitie (Sprint 1)**

* **Literatuurstudie:** Start met het onderzoeken van de meteorologische drijfveren achter droogtes en overstromingen in Afrika.


* **Doelvariabelen bepalen:** Definieer exact wat een "droogte" of "overstroming" is in de context van jullie data en hoe je dit als *target variable* gaat formuleren.



**Fase 2: Data Verzameling & Exploratory Data Analysis (EDA) (Sprint 2)**

* **Omgeving inrichten:** Krijg toegang tot de Microsoft Azure omgeving en zet je Python-ontwikkelomgeving (Jupyter Notebook / VS Code) op met libraries zoals Pandas en Matplotlib.


* **Data verkenning:** Laad de ERA5 data in en analyseer de 6 atmosferische parameters. Controleer op missende waarden, uitschieters en seizoensgebonden patronen.



**Fase 3: Feature Engineering (Sprint 3)**

* **Variabelen transformeren:** Vertaal de ruwe parameters naar bruikbare features voor je model, zoals *rolling rainfall sums* of de duur van droge periodes.


* **Onderbouwing:** Koppel de gekozen features expliciet aan de theorie uit je literatuuronderzoek om je keuzes wetenschappelijk te verantwoorden.



**Fase 4: Modelontwikkeling & Validatie (Sprint 4 & 5)**

* **Baseline model:** Begin simpel met bijvoorbeeld Logistic Regression of Decision Trees.


* **Geavanceerde modellen:** Train complexere modellen zoals Random Forest of Gradient Boosting (XGBoost).


* **Evaluatie:** Test de prestaties van de modellen op robuustheid door gebruik te maken van metrics zoals *precision*, *recall*, *F1-score* en *ROC-AUC*.



**Fase 5: Interpretatie & Ethiek (Sprint 6)**

* **Feature Importance:** Analyseer welke features het model het belangrijkst vindt en controleer of dit overeenkomt met bekende meteorologische mechanismen.


* **Ethische reflectie:** Schrijf de reflectie over mogelijke bias in de data, de impact op lokale boeren, en het verantwoorde gebruik van AI in deze context.



**Fase 6: Oplevering (Sprint 7)**

* **Visualisatie:** Maak de resultaten en voorspellingen inzichtelijk voor non-technische stakeholders (zoals landbouwadviseurs).


* **Afronding:** Documenteer de code netjes, bereid de presentatie voor, en lever het *proof-of-concept* op.
